- Create a list to hold the sequence being read in
- Use a for loop to read the file, line by line
- Capture the sequence name with a conditional
- Before adding each sequence line to the list, remove any newline or space characters

```python
#create a list ot hold the sequence being read in
sequence = []
# use a for loop to read the file, line by line
for line in fs:
    #capture the sequence name with a conditional
    if ">" in line:
        name = line.rstrip("\n") [1:]
        # rstrip strips from the right hand side only because each new line ends wit a newline
        # we can start at position 1 to skip the > that begins each FASTA file
        continue
    # we need to record the information (each line we read) into a variable
    sequence.append(line.rstrip("\n"))

print(name)
print(sequence)
fs.close()
```

- Concatenate the list of sequences into a single sequence string

```python
#join all the strings into the list into one big string
sequence = ''.join(sequence)
print(sequence)

#we just wrote a FASTA file parser in these 8 lines of code
```

Dictionary example

```python
counter = {}
string = 'absgddsag'
for character in string:
    #this only does something when the key hasn't been seen yet and needs to be added to the dictionary
    counter.setdefault(character, 0)
    counter[character] += 1
for key, value in counter.items():
    print(key,value)
```

- Open a filestream to the codon table file
- Create a dictionary to hold the codon/amino acid pairs
- Read in the file, line by line
- Strip newline and space characters from line and split by tabs to create a list of key and value
- Add the key/value pair to the dictionary

```python
fs = open('day2_work/codons.tsv')
codons = {}
for line in fs:
    fields = line.strip().split('\t')
    #split always gives us a list
    codons[fields[0]] = fields[1]
fs.close()
print(codons)
```
